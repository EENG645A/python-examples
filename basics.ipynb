```python
# Important data types: string, list, dictionary
mystring = 'The most basic of strings'
another_string = 'another_string'
print(3)
print(another_string)

# Adding strings actually concatenates them
print(mystring + another_string)

# Python is flexible with types, mostly
print(int(4) + float(3.5))

# this does not work: 
# print(3+another_string)
# but this does:
print(str(3) + another_string)

# 'f strings' are more advanced but very useful for shoving info together. 
# just type lowercase f in front of your string and put variables in {} inside the string!
my_f_string = f'This is my f string containing {another_string} all in one~'
print(my_f_string)
```

```python
# Lists are the most commonly used 'array' in python
empty_list = []
print(empty_list)
# you can add stuff in later with append (which is a built in method for lists)
empty_list.append('stuff')
print(empty_list)

# lists can be sliced like matlab, but are 0 indexed. You can also count from the end of the list starting with -1
longer_list = [0, 1, 2, 3, 4, 5]
print('front of list', longer_list[0])
print('up to 2nd in list', longer_list[:2])
print(f'end of list = {longer_list[-1]}')
print(f'second from end of list = {longer_list[-2]}')
```

```python
# Dictionaries are one of the most useful variable types for storing and accessing more complicated data
# They are in the format {key: data} where key is a string, and data can be pretty much anything.
# They are then indexed by the keys, or by various other functions in python
# This is an empty dictionary:
my_dict = {}

# They may be filled in later like so:
my_dict['mystring'] = 'a string!'
my_dict['myfloat'] = 100. # you may force a float in python by just adding a period like so, or like 100.0
print(my_dict)

# You may also just populate this immediately like so:
my_dict = {'mystring': 'a string!', 'myfloat': 100.0}

# They can even contain other dictionaries, and become many layers deep
# There are many ways to format python, but here is an example of how you might make it more readable using indents:
layers = {
    'my_dict': my_dict,
    'another_dict': {
        'stuff': 'stuff',
        'num': 1e5
    },
    'just a string': 'lonely_string',
    'int': int(1), # trailing commas are okay in python, and I recommend them to avoid forgetting them later
}

print(layers)
```

```python
# Indentation is great to improve readability, but be careful since indents matter in python! 
# Your IDE will usually watch your back and help you not mess this up
# The first time indentations matter is loops. 
# # The indent tells the interpreter what is in the loop, and what is not

# The most basic loop. I don't recommend this, it is not as "pythonic" and prone to mistakes
for i in range(5):
    print(i)
print('Not in the loop')
print() # blank line

# Just iterate directly over your 'iterable' instead
# How about looping over a dictionary instead?
layers = {
    'my_dict': my_dict,
    'another_dict': {
        'stuff': 'stuff',
        'num': 1e5
    },
    'just a string': 'lonely_string',
    'int': int(1), # trailing commas are okay in python, and I recommend them to avoid forgetting them later
}
for layer in layers:
    # this will iterate over each key in the top of the dict as 'layer'
    print(layer) #this is the key (we are iterating over keys)
    print(layers[layer]) #this is the value. There are multiple ways to iter on a dict
    print() # blank line

# HOWEVER! Be careful to not mutate the thing you are iterating on, because that will likely cause undefined behavior.
    # In other words, do NOT do this:
for layer in layers:
    layer = layer+'2' #this might work or it might create hard to trace bugs
    print(layer)
```

```python
# Bonus: what if I need the index in my loop?? use 'enumerate'
my_list = ['stuff', 'things', 'and so on']

for i, item in enumerate(my_list):
    print(f'index is {i}')
    print('item is: ', item, '\n')
```

```python
# If statements use indentation as well
something_true = True
something_false= None
something_list= [1,2,3]

if something_true:
    print('If: something was true')

if something_false:
    print('None was true?')
else:
    print('Else: the none was indeed false')

if {}:
    print('This should not print, empty dicts are false')
elif []:
    print('This should also not print, empty lists are false as well')
elif something_list:
    print('Elif: the not-empty list is true')
else:
    print("we won't get to here because one of the above if / elif were true")

# another thing to note is that numbers are "true" so be careful of syntax
x = 0
if x == 1 or 5:
    print('x is 1 or 5... or is it? How did this if statement return true?')

# because this is correct syntax. The above finds that "5" is true by itself, because an integer is true
if x == 1 or x == 5:
    print('x was 1 or 5')
else:
    print('of course x is not 1 or 5')

if 5:
    print('the integer 5 is true')
if 0:
    print('0 is not true though?')
```

```python
# The next place indents matters is functions and classes
# Python can do object oriented programming!
def myfunc(x,y):
    '''my docstring'''
    return x+y 

print(myfunc(1,2))

# in cases where indentation does not matter, it is useful for code readability
print(myfunc(
    x=2,
    y=5,
))
```

```python
# Run this cell before the below
class MyClass():
    def __init__(self):
        self.mylist = 'a list in the class'

    def concat_with_mylist(self, other_list):
        return self.mylist+other_list

    def reset_to_empty(self):
        self.mylist = ''
```

```python
myobject = MyClass()

print(myobject.mylist)
longer_list = myobject.concat_with_mylist(', and a list from outside')
print(longer_list)

# Now reset to an empty list:
myobject.reset_to_empty()
print(myobject.mylist) # nothing comes out
print('there is an empty list above')
```

```python
# A bit more advanced but very useful
# 'List comprehension' is a very powerful, fast, and compact way to make lists in python using conditions or loops
# Also works with dictionaries!
num_list = [1, 2, 3, 4, 5]
print(num_list)

comprende = [x+1 for x in num_list]
print(comprende)

kaojai = [str(x) for x in num_list]
print(kaojai)

wakatta = [float(x) for x in num_list if x > 1]
print(wakatta)

understood = {f'{x} squared': x**2 for x in num_list}
print(understood)
```
